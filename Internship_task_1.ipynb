**importing libraries used**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

**loading dataset from seaborn library as it is pre installed in seaborn and by deafault it is not a dataframe so then converting it to pandas dataframe for further tasks**

```python
iris = sns.load_dataset('iris')
iris_dataset = pd.DataFrame(iris)
```

```python
iris_dataset.isnull().sum()
```

**Create scatter plots to show relationships between features.**

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.scatterplot(x='sepal_length', y='sepal_width', data=iris_dataset, ax=axes[0])
axes[0].set_title('Sepal Length vs Sepal Width')
sns.scatterplot(x='petal_length', y='petal_width', data=iris_dataset, ax=axes[1])
axes[1].set_title('Petal Length vs Petal Width')
plt.show()
```

**Using histograms to show value distributions.**

```python
iris_dataset.hist(bins=20, figsize=(12, 8))
plt.show()
```

**Using box plots to identify outliers.**

```python
fig,axes = plt.subplots(2, 2, figsize=(12, 8))
axes[0,0].boxplot(iris_dataset['sepal_length'])
axes[0,0].set_title('Boxplot of Sepal Length')
axes[0,1].boxplot(iris_dataset['sepal_width'])
axes[0,1].set_title('Boxplot of Sepal width')
axes[1,0].boxplot(iris_dataset['petal_length'])
axes[1,0].set_title('Boxplot of petal Length')
axes[1,1].boxplot(iris_dataset['petal_width'])
axes[1,1].set_title('Boxplot of petal width')
plt.tight_layout()
plt.show()
```
